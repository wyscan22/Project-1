```python
import sys

class CustomWriter:
    def __init__(self, original_stream):
        self.original_stream = original_stream
    def write(self, text):
        if text.strip():
            self.original_stream.write(text)
        else:
            self.original_stream.write(text)
    def flush(self):
        self.original_stream.flush()

# Context Manager
class ConsoleLogger:
    def __init__(self, file_name):
        assert file_name[-4:] == ".txt" or file_name is None, "File must be a .txt file."

        self.original_stdout = sys.stdout
        self.file_name = file_name
        self.handle = None
        self.current_writer = None

    def __enter__(self):
        print(f"Entering <{self.file_name}>...")

        if self.file_name:
            self.handle = open(self.file_name, "w")
            self.current_writer = CustomWriter(self.handle)
        else:
            self.current_writer = sys.stdout
        sys.stdout = self.current_writer

        return self
    def __exit__(self, exc_type, exc_val, exc_tb):
        sys.stdout = self.original_stdout

        if exc_type:
            self.handle.write(f"Error: {exc_type.__name__} ({exc_val})\n")
            return True

        if self.handle: self.handle.close()
        print(f"Exiting <{self.file_name}>...")
        return False

with ConsoleLogger("ConsoleLog.txt") as Console:
    print("em pee kay")
    print("gee pee kay")
    2 / 0

print("em pee kay")
```
