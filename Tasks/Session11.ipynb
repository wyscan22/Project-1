```python
t = time.time()

def remove_suffix(string, suffix):
    if string[-len(suffix):] == suffix:
        return string[:-len(suffix)]
    else: return string
def remove_prefix(string, prefix):
    if string[:len(prefix)] == prefix:
        return string[len(prefix):]
    else: return string

print(remove_prefix("mpkpk", "m"))

print("\033[37m" + display_ETime(time.time() - t) + "\033[0m")
```

```python
t = time.time()

# [PJT] = "C:/Users/ASUS/PycharmProjects/Project-1"

def parse_shebang(file_name):
    handle = open(file_name, "r")
    result = handle.read().split("\n")[0]
    if result[:2] == "#!":
        handle.close()
        result = remove_prefix(result, "#!")
        while result[0] == " ":
            result = remove_prefix(result, " ")
    else: return None
    handle.close()
    return result

print("\033[37m" + display_ETime(time.time() - t) + "\033[0m")
```
